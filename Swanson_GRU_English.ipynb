
# Swanson Stream GRU — High-Water-Risk Classification

This notebook develops a **Swanson Stream-specific GRU** using the same methodology applied to Mahurangi and Otara.

### Objective
Use the previous **48 hours** of:

- `rainfall_mm`
- `discharge_m3s`
- `water_level_m`

to estimate whether the water level will exceed a **catchment-specific high-water-risk threshold during the next 24 hours**.

### Methodological consistency
- Chronological train / validation / test split.
- High-water threshold = **99.5th percentile of training-period water level only**.
- 48-hour GRU input window.
- 24-hour forecast horizon.
- Missing input values are not interpolated; invalid sequences are excluded.
- Class imbalance is handled with `pos_weight`.
- Probability threshold is selected on validation data using F2-score.
- Test data remain untouched until final evaluation.

The target is a **statistical high-water-risk proxy**, not an independently verified historical flood label.


## Fair Comparison with LSTM

This GRU experiment intentionally keeps the **same modelling protocol as the corresponding LSTM**:

- same catchment data and target;
- same chronological Training / Validation / Test periods;
- same three inputs: `rainfall_mm`, `discharge_m3s`, and `water_level_m`;
- same 48-hour lookback;
- same 24-hour high-water-risk prediction horizon;
- same training-only scaling;
- same class-imbalance weighting;
- same hidden size (32), dense layer (16), dropout (0.25), optimizer, learning rate, batch size, early stopping, and validation-based probability-threshold selection;
- same evaluation metrics and episode-level diagnostic.

The **recurrent unit is the main architectural change**: `nn.LSTM` is replaced by `nn.GRU`. This makes the GRU-vs-LSTM comparison substantially more controlled.

```python

import copy
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    precision_recall_curve,
    precision_score,
    recall_score,
    f1_score,
    fbeta_score,
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    ConfusionMatrixDisplay,
    RocCurveDisplay,
    PrecisionRecallDisplay,
)

import joblib
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
```

## 2. Load Swanson Water-Level/Rainfall Data and Discharge Station 7907

```python

swanson_path = Path("swanson_cleaned_hourly.csv")
discharge_path = Path("7907.csv")

if not swanson_path.exists():
    raise FileNotFoundError(
        "swanson_cleaned_hourly.csv was not found. "
        "Place it in the same folder as this notebook."
    )

if not discharge_path.exists():
    raise FileNotFoundError(
        "7907.csv was not found. "
        "Place it in the same folder as this notebook."
    )

swanson_raw = pd.read_csv(swanson_path)

required_swanson_columns = [
    "start_time",
    "rainfall_mm",
    "water_level_m",
]

missing_swanson_columns = [
    col for col in required_swanson_columns
    if col not in swanson_raw.columns
]

if missing_swanson_columns:
    raise ValueError(
        f"Missing required Swanson columns: {missing_swanson_columns}"
    )

swanson_base = swanson_raw[
    required_swanson_columns
].copy()

swanson_base["start_time"] = pd.to_datetime(
    swanson_base["start_time"],
    errors="coerce"
)

# The supplied cleaned file contains one duplicate midnight row
# on 1 January in several years. The duplicated core measurements
# are identical, so one copy is retained.
duplicate_count = swanson_base["start_time"].duplicated().sum()

swanson_base = (
    swanson_base
    .drop_duplicates(
        subset=["start_time"],
        keep="first"
    )
    .sort_values("start_time")
    .reset_index(drop=True)
)

print("Duplicate timestamps removed:", duplicate_count)

# The 7907 export has a metadata row before its true header.
discharge = pd.read_csv(
    discharge_path,
    skiprows=1
)

discharge = discharge.iloc[:, :3].copy()
discharge.columns = [
    "start_time",
    "end_time",
    "discharge_m3s",
]

discharge["start_time"] = pd.to_datetime(
    discharge["start_time"],
    errors="coerce"
)

discharge["discharge_m3s"] = pd.to_numeric(
    discharge["discharge_m3s"],
    errors="coerce"
)

discharge = (
    discharge[
        [
            "start_time",
            "discharge_m3s"
        ]
    ]
    .drop_duplicates(
        subset=["start_time"],
        keep="first"
    )
)

df = (
    swanson_base
    .merge(
        discharge,
        on="start_time",
        how="left"
    )
    .rename(
        columns={
            "start_time": "timestamp"
        }
    )
    [
        [
            "timestamp",
            "rainfall_mm",
            "discharge_m3s",
            "water_level_m",
        ]
    ]
    .sort_values("timestamp")
    .reset_index(drop=True)
)

display(df.head())
```

### Combined Swanson Inputs

Swanson requires rainfall/water-level data from `swanson_cleaned_hourly.csv` and discharge from station `7907.csv`.

The two sources are merged **in memory** for the GRU. No additional consolidated CSV is exported here because the GRU uses the same underlying inputs already prepared for the previous modelling workflow.

## 3. Audit the Combined Hourly Series

```python

print("Shape:", df.shape)

print("\nDate range:")
print(df["timestamp"].min(), "to", df["timestamp"].max())

print("\nDuplicate timestamps:")
print(df["timestamp"].duplicated().sum())

print("\nMissing values:")
print(df.isna().sum())

time_difference = df["timestamp"].diff()

non_hourly_gap = (
    time_difference.notna()
    & (
        time_difference
        != pd.Timedelta(hours=1)
    )
)

print("\nNumber of non-hourly gaps:")
print(non_hourly_gap.sum())

if non_hourly_gap.sum() > 0:
    display(
        df.loc[
            non_hourly_gap,
            ["timestamp"]
        ].head(20)
    )
```


## 4. Chronological Train / Validation / Test Split

The cleaned Swanson series runs continuously from early 2015 to the end of 2025.

The same 70% / 15% / 15% chronological split strategy used for the other catchments is applied here.


```python

model_data = df.copy()

number_of_rows = len(model_data)

train_end = int(number_of_rows * 0.70)
validation_end = int(number_of_rows * 0.85)

model_data["data_period"] = "test"

model_data.loc[
    model_data.index < train_end,
    "data_period"
] = "train"

model_data.loc[
    (model_data.index >= train_end)
    & (model_data.index < validation_end),
    "data_period"
] = "validation"

split_summary = (
    model_data
    .groupby("data_period")
    .agg(
        rows=("timestamp", "size"),
        beginning=("timestamp", "min"),
        ending=("timestamp", "max"),
    )
    .loc[
        ["train", "validation", "test"]
    ]
)

display(split_summary)
```

## 5. Define the Statistical High-Water-Risk Target

```python

HIGH_WATER_PERCENTILE = 0.995

high_water_threshold = (
    model_data.loc[
        model_data["data_period"] == "train",
        "water_level_m"
    ]
    .dropna()
    .quantile(HIGH_WATER_PERCENTILE)
)

print(
    "Training-only 99.5th percentile high-water threshold:",
    round(high_water_threshold, 3),
    "m"
)
```

```python

FORECAST_HORIZON = 24

future_water_levels = pd.concat(
    [
        model_data["water_level_m"].shift(-hour)
        for hour in range(
            1,
            FORECAST_HORIZON + 1
        )
    ],
    axis=1
)

# Require every future water-level observation to be available.
model_data["future_24h_max_water_level"] = (
    future_water_levels.max(
        axis=1,
        skipna=False
    )
)

model_data["flood_next_24h"] = np.where(
    model_data[
        "future_24h_max_water_level"
    ].isna(),
    np.nan,
    (
        model_data[
            "future_24h_max_water_level"
        ]
        >= high_water_threshold
    ).astype(int)
)

# Prevent a target from crossing from one data period into another.
future_periods = pd.concat(
    [
        model_data["data_period"].shift(-hour)
        for hour in range(
            1,
            FORECAST_HORIZON + 1
        )
    ],
    axis=1
)

same_period_target = future_periods.eq(
    model_data["data_period"],
    axis=0
).all(axis=1)

model_data.loc[
    ~same_period_target,
    [
        "future_24h_max_water_level",
        "flood_next_24h",
    ]
] = np.nan

print(
    "Rows with unavailable or boundary-crossing targets:",
    model_data["flood_next_24h"].isna().sum()
)
```

```python

target_summary = (
    model_data
    .dropna(
        subset=["flood_next_24h"]
    )
    .groupby("data_period")[
        "flood_next_24h"
    ]
    .agg(
        total_rows="count",
        high_water_risk_hours="sum",
        high_water_risk_percentage="mean",
    )
    .loc[
        ["train", "validation", "test"]
    ]
)

target_summary[
    "high_water_risk_percentage"
] *= 100

display(
    target_summary.round(3)
)
```


## 6. Missing Input Data

No interpolation is performed.

A sequence is valid only when all 48 historical hours contain rainfall, discharge, and water-level measurements.


```python

feature_columns = [
    "rainfall_mm",
    "discharge_m3s",
    "water_level_m",
]

model_data["features_available"] = (
    model_data[
        feature_columns
    ]
    .notna()
    .all(axis=1)
)

missing_rows = (
    ~model_data["features_available"]
)

print(
    "Rows with at least one missing input:",
    missing_rows.sum()
)

missing_groups = (
    missing_rows
    .ne(
        missing_rows.shift(
            fill_value=False
        )
    )
    .cumsum()
)

missing_episodes = (
    model_data.loc[missing_rows]
    .assign(
        missing_group=missing_groups[
            missing_rows
        ]
    )
    .groupby("missing_group")
    .agg(
        start=("timestamp", "min"),
        end=("timestamp", "max"),
        hours=("timestamp", "size"),
    )
    .reset_index(drop=True)
)

print(
    "Missing-data episodes:",
    len(missing_episodes)
)

if len(missing_episodes) > 0:
    print(
        "Longest missing-data episode:",
        missing_episodes["hours"].max(),
        "hours"
    )

    display(
        missing_episodes.head(20)
    )
```

## 7. GRU Lookback

```python

LOOKBACK_HOURS = 48

candidate_lookbacks = [
    24,
    48,
    72,
]

lookback_results = []

for lookback in candidate_lookbacks:

    valid_input_sequence = (
        model_data[
            "features_available"
        ]
        .astype(int)
        .rolling(
            window=lookback,
            min_periods=lookback
        )
        .sum()
        .eq(lookback)
    )

    valid_sample = (
        valid_input_sequence
        & model_data[
            "flood_next_24h"
        ].notna()
    )

    for period in [
        "train",
        "validation",
        "test"
    ]:

        lookback_results.append({
            "lookback_hours": lookback,
            "data_period": period,
            "valid_samples": int(
                (
                    valid_sample
                    & model_data[
                        "data_period"
                    ].eq(period)
                ).sum()
            )
        })

lookback_summary = pd.DataFrame(
    lookback_results
)

display(
    lookback_summary.pivot(
        index="lookback_hours",
        columns="data_period",
        values="valid_samples"
    )
)
```

## 8. Scale Inputs Using Training Data Only

```python

scaler_fit_mask = (
    model_data[
        "data_period"
    ].eq("train")
    & model_data[
        "features_available"
    ]
)

scaler = StandardScaler()

scaler.fit(
    model_data.loc[
        scaler_fit_mask,
        feature_columns
    ]
)

scaled_features = scaler.transform(
    model_data[
        feature_columns
    ]
).astype(np.float32)

print("Training means:")
display(
    pd.Series(
        scaler.mean_,
        index=feature_columns
    ).to_frame(
        "mean"
    )
)

print("\nTraining scales:")
display(
    pd.Series(
        scaler.scale_,
        index=feature_columns
    ).to_frame(
        "scale"
    )
)
```

## 9. Convert the Hourly Series into 3D GRU Sequences

```python

from numpy.lib.stride_tricks import sliding_window_view

def create_sequences(
    data,
    scaled_feature_array,
    lookback,
    period_name,
):
    targets = (
        data[
            "flood_next_24h"
        ]
        .to_numpy()
    )

    periods = (
        data[
            "data_period"
        ]
        .to_numpy()
    )

    feature_available = (
        np.isfinite(
            scaled_feature_array
        )
        .all(axis=1)
    )

    complete_window = (
        pd.Series(
            feature_available.astype(int)
        )
        .rolling(
            window=lookback,
            min_periods=lookback
        )
        .sum()
        .to_numpy()
        == lookback
    )

    valid_end = (
        complete_window
        & np.isfinite(targets)
        & (
            periods
            == period_name
        )
    )

    end_indices = np.where(
        valid_end
    )[0]

    all_windows = (
        sliding_window_view(
            scaled_feature_array,
            window_shape=lookback,
            axis=0
        )
        .transpose(
            0,
            2,
            1
        )
    )

    window_positions = (
        end_indices
        - (lookback - 1)
    )

    X = (
        all_windows[
            window_positions
        ]
        .copy()
        .astype(
            np.float32
        )
    )

    y = (
        targets[
            end_indices
        ]
        .astype(
            np.float32
        )
    )

    timestamps = (
        data.loc[
            end_indices,
            "timestamp"
        ]
        .to_numpy()
    )

    return (
        X,
        y,
        timestamps,
        end_indices,
    )
```

```python

X_train, y_train, train_timestamps, train_indices = (
    create_sequences(
        model_data,
        scaled_features,
        LOOKBACK_HOURS,
        "train"
    )
)

X_validation, y_validation, validation_timestamps, validation_indices = (
    create_sequences(
        model_data,
        scaled_features,
        LOOKBACK_HOURS,
        "validation"
    )
)

print(
    "X_train:",
    X_train.shape
)
print(
    "y_train:",
    y_train.shape
)
print(
    "X_validation:",
    X_validation.shape
)
print(
    "y_validation:",
    y_validation.shape
)

print("\nPositive percentage:")
print(
    "Train:",
    round(
        y_train.mean() * 100,
        3
    ),
    "%"
)
print(
    "Validation:",
    round(
        y_validation.mean() * 100,
        3
    ),
    "%"
)
```

## 10. Handle Class Imbalance

```python

negative_count = int(
    (y_train == 0).sum()
)

positive_count = int(
    (y_train == 1).sum()
)

positive_weight = (
    negative_count
    / positive_count
)

print(
    "Negative samples:",
    negative_count
)

print(
    "Positive samples:",
    positive_count
)

print(
    "Positive class weight:",
    round(
        positive_weight,
        3
    )
)
```

## 11. GRU Architecture

```python

class GRUClassifier(
    nn.Module
):
    def __init__(
        self,
        input_size=3,
        hidden_size=32,
        dense_size=16,
        dropout=0.25,
    ):
        super().__init__()

        self.gru = nn.GRU(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True,
        )

        self.dropout = (
            nn.Dropout(
                dropout
            )
        )

        self.dense = nn.Linear(
            hidden_size,
            dense_size
        )

        self.relu = nn.ReLU()

        self.output = nn.Linear(
            dense_size,
            1
        )

    def forward(
        self,
        x
    ):
        _, hidden_state = self.gru(x)

        x = hidden_state[-1]

        x = self.dropout(x)
        x = self.dense(x)
        x = self.relu(x)

        logits = self.output(x)

        return logits.squeeze(1)


model = GRUClassifier(
    input_size=len(
        feature_columns
    ),
    hidden_size=32,
    dense_size=16,
    dropout=0.25,
)

print(model)
```

## 12. Train the GRU

```python

torch.manual_seed(SEED)
np.random.seed(SEED)

BATCH_SIZE = 512
MAX_EPOCHS = 15
PATIENCE = 4
LEARNING_RATE = 0.001
MIN_DELTA = 1e-4

model = GRUClassifier(
    input_size=len(
        feature_columns
    ),
    hidden_size=32,
    dense_size=16,
    dropout=0.25,
)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE
)

loss_function = (
    nn.BCEWithLogitsLoss(
        pos_weight=torch.tensor(
            positive_weight,
            dtype=torch.float32
        )
    )
)

train_loader = DataLoader(
    TensorDataset(
        torch.from_numpy(
            X_train
        ),
        torch.from_numpy(
            y_train
        )
    ),
    batch_size=BATCH_SIZE,
    shuffle=False,
)

validation_loader = DataLoader(
    TensorDataset(
        torch.from_numpy(
            X_validation
        ),
        torch.from_numpy(
            y_validation
        )
    ),
    batch_size=1024,
    shuffle=False,
)

best_validation_pr_auc = -np.inf
best_state = None
best_epoch = None
epochs_without_improvement = 0
training_history = []

for epoch in range(
    1,
    MAX_EPOCHS + 1
):

    model.train()

    total_loss = 0.0

    for (
        X_batch,
        y_batch
    ) in train_loader:

        optimizer.zero_grad()

        logits = model(
            X_batch
        )

        loss = loss_function(
            logits,
            y_batch
        )

        loss.backward()
        optimizer.step()

        total_loss += (
            loss.item()
            * len(y_batch)
        )

    epoch_loss = (
        total_loss
        / len(y_train)
    )

    model.eval()

    validation_probability_parts = []

    with torch.no_grad():

        for (
            X_batch,
            _
        ) in validation_loader:

            probabilities = torch.sigmoid(
                model(
                    X_batch
                )
            )

            validation_probability_parts.append(
                probabilities.cpu().numpy()
            )

    validation_probabilities = np.concatenate(
        validation_probability_parts
    )

    validation_pr_auc = (
        average_precision_score(
            y_validation,
            validation_probabilities
        )
    )

    validation_roc_auc = (
        roc_auc_score(
            y_validation,
            validation_probabilities
        )
    )

    training_history.append({
        "epoch": epoch,
        "train_loss": epoch_loss,
        "validation_pr_auc": validation_pr_auc,
        "validation_roc_auc": validation_roc_auc,
    })

    print(
        f"Epoch {epoch:02d} | "
        f"loss={epoch_loss:.4f} | "
        f"val_PR_AUC={validation_pr_auc:.4f} | "
        f"val_ROC_AUC={validation_roc_auc:.4f}"
    )

    if (
        validation_pr_auc
        > best_validation_pr_auc
        + MIN_DELTA
    ):
        best_validation_pr_auc = (
            validation_pr_auc
        )

        best_state = copy.deepcopy(
            model.state_dict()
        )

        best_epoch = epoch

        epochs_without_improvement = 0

    else:
        epochs_without_improvement += 1

        if (
            epochs_without_improvement
            >= PATIENCE
        ):
            print(
                "\nEarly stopping triggered."
            )
            break

model.load_state_dict(
    best_state
)

history_df = pd.DataFrame(
    training_history
)

print(
    "\nBest epoch:",
    best_epoch
)

print(
    "Best validation PR-AUC:",
    round(
        best_validation_pr_auc,
        4
    )
)
```

## 13. Select the Probability Threshold on Validation Data

```python

model.eval()

with torch.no_grad():

    validation_probabilities = torch.sigmoid(
        model(
            torch.from_numpy(
                X_validation
            )
        )
    ).cpu().numpy()

(
    precision_curve,
    recall_curve,
    threshold_curve,
) = precision_recall_curve(
    y_validation,
    validation_probabilities
)

precision_for_threshold = (
    precision_curve[:-1]
)

recall_for_threshold = (
    recall_curve[:-1]
)

f2_curve = (
    5
    * precision_for_threshold
    * recall_for_threshold
    / (
        4
        * precision_for_threshold
        + recall_for_threshold
        + 1e-12
    )
)

best_threshold_position = (
    np.argmax(
        f2_curve
    )
)

selected_probability_threshold = float(
    threshold_curve[
        best_threshold_position
    ]
)

validation_predictions = (
    validation_probabilities
    >= selected_probability_threshold
).astype(int)

tn_val, fp_val, fn_val, tp_val = (
    confusion_matrix(
        y_validation,
        validation_predictions
    )
    .ravel()
)

validation_metrics = pd.Series({
    "threshold": selected_probability_threshold,
    "accuracy": accuracy_score(
        y_validation,
        validation_predictions
    ),
    "precision": precision_score(
        y_validation,
        validation_predictions,
        zero_division=0
    ),
    "recall": recall_score(
        y_validation,
        validation_predictions,
        zero_division=0
    ),
    "f1_score": f1_score(
        y_validation,
        validation_predictions,
        zero_division=0
    ),
    "f2_score": fbeta_score(
        y_validation,
        validation_predictions,
        beta=2,
        zero_division=0
    ),
    "roc_auc": roc_auc_score(
        y_validation,
        validation_probabilities
    ),
    "pr_auc": average_precision_score(
        y_validation,
        validation_probabilities
    ),
    "true_positives": tp_val,
    "false_negatives": fn_val,
    "false_positives": fp_val,
    "true_negatives": tn_val,
})

display(
    validation_metrics
    .to_frame(
        "Validation"
    )
    .round(4)
)
```

## 14. Final Evaluation on the Untouched Test Set

```python

X_test, y_test, test_timestamps, test_indices = (
    create_sequences(
        model_data,
        scaled_features,
        LOOKBACK_HOURS,
        "test"
    )
)

print(
    "X_test:",
    X_test.shape
)

print(
    "y_test:",
    y_test.shape
)
```

```python

model.eval()

test_loader = DataLoader(
    TensorDataset(
        torch.from_numpy(
            X_test
        ),
        torch.from_numpy(
            y_test
        )
    ),
    batch_size=1024,
    shuffle=False,
)

test_probability_parts = []

with torch.no_grad():

    for (
        X_batch,
        _
    ) in test_loader:

        probabilities = torch.sigmoid(
            model(
                X_batch
            )
        )

        test_probability_parts.append(
            probabilities.cpu().numpy()
        )

test_probabilities = np.concatenate(
    test_probability_parts
)

test_predictions_binary = (
    test_probabilities
    >= selected_probability_threshold
).astype(int)

tn, fp, fn, tp = (
    confusion_matrix(
        y_test,
        test_predictions_binary
    )
    .ravel()
)

test_metrics = pd.Series({
    "selected_threshold": selected_probability_threshold,
    "accuracy": accuracy_score(
        y_test,
        test_predictions_binary
    ),
    "precision": precision_score(
        y_test,
        test_predictions_binary,
        zero_division=0
    ),
    "recall": recall_score(
        y_test,
        test_predictions_binary,
        zero_division=0
    ),
    "f1_score": f1_score(
        y_test,
        test_predictions_binary,
        zero_division=0
    ),
    "f2_score": fbeta_score(
        y_test,
        test_predictions_binary,
        beta=2,
        zero_division=0
    ),
    "roc_auc": roc_auc_score(
        y_test,
        test_probabilities
    ),
    "pr_auc": average_precision_score(
        y_test,
        test_probabilities
    ),
    "true_positives": tp,
    "false_negatives": fn,
    "false_positives": fp,
    "true_negatives": tn,
})

display(
    test_metrics
    .to_frame(
        "Test"
    )
    .round(4)
)

print("\nClassification report:")

print(
    classification_report(
        y_test,
        test_predictions_binary,
        digits=4,
        zero_division=0
    )
)
```


## 17. Episode-Level Diagnostic Evaluation

Consecutive positive target hours are grouped into model-defined high-water-risk episodes.

An episode is counted as detected if the GRU produces at least one positive prediction during that episode.

This is a diagnostic measure of **high-water-risk episode detection**, not verification of officially recorded historical floods.


```python

full_test_target = (
    model_data.loc[
        model_data[
            "data_period"
        ].eq("test")
        & model_data[
            "flood_next_24h"
        ].notna(),
        [
            "timestamp",
            "flood_next_24h",
        ]
    ]
    .rename(
        columns={
            "flood_next_24h":
            "actual_risk"
        }
    )
    .copy()
)

full_test_target["actual_risk"] = (
    full_test_target[
        "actual_risk"
    ]
    .astype(int)
)

event_frame = (
    full_test_target
    .merge(
        test_predictions[
            [
                "timestamp",
                "risk_probability",
                "predicted_risk",
            ]
        ],
        on="timestamp",
        how="left"
    )
    .sort_values(
        "timestamp"
    )
    .reset_index(
        drop=True
    )
)

event_frame[
    "predicted_risk_for_detection"
] = (
    event_frame[
        "predicted_risk"
    ]
    .fillna(0)
    .astype(int)
)

actual_positive = (
    event_frame[
        "actual_risk"
    ]
    == 1
)

timestamp_gap = (
    event_frame[
        "timestamp"
    ]
    .diff()
    .ne(
        pd.Timedelta(
            hours=1
        )
    )
)

new_event = (
    actual_positive
    & (
        ~actual_positive.shift(
            fill_value=False
        )
        | timestamp_gap
    )
)

event_frame[
    "event_id"
] = (
    new_event
    .cumsum()
    .where(
        actual_positive
    )
)

event_level_results = (
    event_frame.loc[
        actual_positive
    ]
    .groupby(
        "event_id"
    )
    .agg(
        event_start=(
            "timestamp",
            "min"
        ),
        event_end=(
            "timestamp",
            "max"
        ),
        event_duration_hours=(
            "timestamp",
            "size"
        ),
        detected_hours=(
            "predicted_risk_for_detection",
            "sum"
        ),
        maximum_risk_score=(
            "risk_probability",
            "max"
        ),
        mean_risk_score=(
            "risk_probability",
            "mean"
        ),
    )
    .reset_index()
)

event_level_results[
    "event_detected"
] = (
    event_level_results[
        "detected_hours"
    ]
    > 0
)

event_level_results[
    "hourly_detection_percentage"
] = (
    100
    * event_level_results[
        "detected_hours"
    ]
    / event_level_results[
        "event_duration_hours"
    ]
)

display(
    event_level_results
)

event_detection_rate = (
    event_level_results[
        "event_detected"
    ]
    .mean()
    * 100
)

print(
    "\nEvent detection rate:",
    round(
        event_detection_rate,
        2
    ),
    "%"
)
```

## GRU vs LSTM Interpretation Reminder

Do not decide whether GRU is better from accuracy alone. The main comparison should use **Validation F2** for model selection, with Recall, Precision, PR-AUC, ROC-AUC, and Test generalisation used as supporting evidence.

A higher GRU score does not automatically mean GRU is universally better; it means it performed better for that catchment under the same modelling protocol.
